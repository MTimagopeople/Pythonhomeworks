```python
import pyodbc

import pandas as pd

# Database connection
conn = pyodbc.connect(
    "DRIVER={SQL Server};"
    "SERVER=62d8c121b25f;"  # Double backslash for escaping
    "DATABASE=Demo_database;"  
    "Trusted_Connection=yes;"
)
cursor = conn.cursor()

# Function to insert data from CSV into SQL Server
def insert_data_from_csv(table_name, csv_file, columns):
    df = pd.read_csv(csv_file, names=columns)
    for index, row in df.iterrows():
        cursor.execute(
            f"INSERT INTO {table_name} ({', '.join(columns)}) VALUES ({', '.join(['?' for _ in columns])})",
            tuple(row)
        )
    conn.commit()

# Insert data into SQL Server tables
insert_data_from_csv("Customers", "Customers.csv", ['CustomerID', 'FullName', 'DOB', 'Email', 'PhoneNumber', 'Address', 'NationalID', 'TaxID', 'EmploymentStatus', 'AnnualIncome', 'CreatedAt', 'UpdatedAt'])
insert_data_from_csv("Branches", "Branches.csv", ['BranchID', 'BranchName', 'Address', 'City', 'State', 'Country', 'ManagerID', 'ContactNumber'])
insert_data_from_csv("Departments", "Departments.csv", ['DepartmentID', 'DepartmentName', 'ManagerID'])
insert_data_from_csv("Employees", "Employees.csv", ['EmployeeID', 'BranchID', 'FullName', 'Position', 'DepartmentID', 'Salary', 'HireDate', 'Status'])
insert_data_from_csv("Accounts", "Accounts.csv", ['AccountID', 'CustomerID', 'AccountType', 'Balance', 'Currency', 'Status', 'BranchID', 'CreatedDate'])
insert_data_from_csv("Transactions", "Transactions.csv", ['TransactionID', 'AccountID', 'TransactionType', 'Amount', 'Currency', 'Date', 'Status', 'ReferenceNo', 'State'])
insert_data_from_csv("CreditCards", "CreditCards.csv", ['CardID', 'CustomerID', 'CardNumber', 'CardType', 'CVV', 'ExpiryDate', 'Limit', 'Status'])
insert_data_from_csv("Merchants", "Merchants.csv", ['MerchantID', 'MerchantName', 'Industry', 'Location', 'CustomerID'])
insert_data_from_csv("CreditCardTransactions", "CreditCardTransactions.csv", ['TransactionID', 'CardID', 'MerchantID', 'Amount', 'Currency', 'Date', 'Status', 'State'])
insert_data_from_csv("OnlineBankingUsers", "OnlineBankingUsers.csv", ['UserID', 'CustomerID', 'Username', 'PasswordHash', 'LastLogin'])
insert_data_from_csv("BillPayments", "BillPayments.csv", ['PaymentID', 'CustomerID', 'BillerName', 'Amount', 'Date', 'Status'])
insert_data_from_csv("MobileBankingTransactions", "MobileBankingTransactions.csv", ['TransactionID', 'CustomerID', 'DeviceID', 'AppVersion', 'TransactionType', 'Amount', 'Date', 'State'])
insert_data_from_csv("Loans", "Loans.csv", ['LoanID', 'CustomerID', 'LoanType', 'Amount', 'InterestRate', 'StartDate', 'EndDate', 'Status'])
insert_data_from_csv("LoanPayments", "LoanPayments.csv", ['PaymentID', 'LoanID', 'AmountPaid', 'PaymentDate', 'RemainingBalance'])
insert_data_from_csv("CreditScores", "CreditScores.csv", ['CustomerID', 'CreditScore', 'UpdatedAt'])
insert_data_from_csv("DebtCollection", "DebtCollection.csv", ['DebtID', 'CustomerID', 'AmountDue', 'DueDate', 'CollectorAssigned'])
insert_data_from_csv("KYC", "KYC.csv", ['KYCID', 'CustomerID', 'DocumentType', 'DocumentNumber', 'VerifiedBy'])
insert_data_from_csv("FraudDetection", "FraudDetection.csv", ['FraudID', 'CustomerID', 'TransactionID', 'RiskLevel', 'ReportedDate'])
insert_data_from_csv("AMLCases", "AMLCases.csv", ['CaseID', 'CustomerID', 'CaseType', 'Status', 'InvestigatorID'])
insert_data_from_csv("RegulatoryReports", "RegulatoryReports.csv", ['ReportID', 'ReportType', 'SubmissionDate'])
insert_data_from_csv("Salaries", "Salaries.csv", ['SalaryID', 'EmployeeID', 'BaseSalary', 'Bonus', 'Deductions', 'PaymentDate'])
insert_data_from_csv("EmployeeAttendance", "EmployeeAttendance.csv", ['AttendanceID', 'EmployeeID', 'CheckInTime', 'CheckOutTime', 'TotalHours'])
insert_data_from_csv("Investments", "Investments.csv", ['InvestmentID', 'CustomerID', 'InvestmentType', 'Amount', 'ROI', 'MaturityDate'])
insert_data_from_csv("StockTradingAccounts", "StockTradingAccounts.csv", ['AccountID', 'CustomerID', 'BrokerageFirm', 'TotalInvested', 'CurrentValue'])
insert_data_from_csv("ForeignExchange", "ForeignExchange.csv", ['FXID', 'CustomerID', 'CurrencyPair', 'ExchangeRate', 'AmountExchanged'])
insert_data_from_csv("InsurancePolicies", "InsurancePolicies.csv", ['PolicyID', 'CustomerID', 'InsuranceType', 'PremiumAmount', 'CoverageAmount'])
insert_data_from_csv("Claims", "Claims.csv", ['ClaimID', 'PolicyID', 'ClaimAmount', 'Status', 'FiledDate'])
insert_data_from_csv("UserAccessLogs", "UserAccessLogs.csv", ['LogID', 'UserID', 'ActionType', 'Timestamp'])
insert_data_from_csv("CyberSecurityIncidents", "CyberSecurityIncidents.csv", ['IncidentID', 'AffectedSystem', 'ReportedDate', 'ResolutionStatus'])
insert_data_from_csv("MerchantTransactions", "MerchantTransactions.csv", ['TransactionID', 'MerchantID', 'Amount', 'PaymentMethod', 'Date', 'State'])

# Close the cursor and connection
cursor.close()
conn.close()

print("Data has been successfully inserted into the SQL Server database.")
```
