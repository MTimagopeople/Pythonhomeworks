```python
import pandas as pd

df = pd.read_csv('task/sales_data.csv')

category_stats = df.groupby('Category').agg(
    total_quantity_sold=('Quantity', 'sum'),
    average_price_per_unit=('Price', 'mean'),
    max_quantity_sold=('Quantity', 'max')
).reset_index()

print("Aggregate statistics for each category:")
print(category_stats)

top_selling_products = df.groupby(['Category', 'Product']).agg(
    total_quantity_sold=('Quantity', 'sum')
).reset_index()

top_selling_products = top_selling_products.loc[top_selling_products.groupby('Category')['total_quantity_sold'].idxmax()]

print("\nTop-selling product in each category:")
print(top_selling_products)

df['Total_Sales'] = df['Quantity'] * df['Price']

highest_sales_date = df.groupby('Date').agg(
    total_sales=('Total_Sales', 'sum')
).reset_index()

highest_sales_date = highest_sales_date.loc[highest_sales_date['total_sales'].idxmax()]

print("\nDate with the highest total sales:")
print(highest_sales_date[['Date', 'total_sales']])
```

```python
import pandas as pd

df = pd.read_csv('task/customer_orders.csv')

customer_order_counts = df.groupby('CustomerID').agg(
    total_orders=('OrderID', 'count')
).reset_index()

customers_with_20_orders = customer_order_counts[customer_order_counts['total_orders'] >= 20]

print("Customers with 20 or more orders:")
print(customers_with_20_orders)

average_price_per_customer = df.groupby('CustomerID').agg(
    average_price=('Price', 'mean')
).reset_index()

customers_with_high_avg_price = average_price_per_customer[average_price_per_customer['average_price'] > 120]

print("\nCustomers with an average price per unit greater than $120:")
print(customers_with_high_avg_price)

product_totals = df.groupby('Product').agg(
    total_quantity=('Quantity', 'sum'),
    total_price=('Price', 'sum')
).reset_index()

products_with_min_quantity = product_totals[product_totals['total_quantity'] >= 5]

print("\nProducts with a total quantity of 5 or more units:")
print(products_with_min_quantity)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('task/population.db')

df = pd.read_sql_query("SELECT * FROM population", conn)

salary_categories = pd.read_excel('task/population salary analysis.xlsx')

df = pd.merge(df, salary_categories, on='salary_band', how='left')

category_stats = df.groupby('salary_band').agg(
    population_count=('population', 'count'),
    average_salary=('salary', 'mean'),
    median_salary=('salary', 'median')
).reset_index()

category_stats['population_percentage'] = (category_stats['population_count'] / category_stats['population_count'].sum()) * 100

print("Measures for each salary category:")
print(category_stats)

state_category_stats = df.groupby(['state', 'salary_band']).agg(
    population_count=('population', 'count'),
    average_salary=('salary', 'mean'),
    median_salary=('salary', 'median')
).reset_index()

state_totals = df.groupby('state').agg(
    total_population=('population', 'count')
).reset_index()

state_category_stats = pd.merge(state_category_stats, state_totals, on='state')
state_category_stats['population_percentage'] = (state_category_stats['population_count'] / state_category_stats['total_population']) * 100

print("\nMeasures for each salary category in each state:")
print(state_category_stats)

conn.close()
```
