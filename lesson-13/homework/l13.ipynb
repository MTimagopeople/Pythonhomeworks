```python
date_string = '2025.02_07 15:15:43'
parsed_date = datetime.datetime.strptime(date_string, '%Y.%m_%d %H:%M:%S')
print(parsed_date)
print(type(parsed_date))
formatted_date = datetime.datetime.strftime(parsed_date, '%A, %B %d, %Y')
print(formatted_date)
```

## Re (RegEx)

```python
import re
```

```python
pattern = 'he'
word = 'hello, team'
result = re.match(pattern, word)
print(result.group())
```

```python
pattern = 'h'
word = 'hello, hey'
result = re.search(pattern, word)
if result:
    print(result.group())
else:
    print('No match')
```

```python
pattern = r'a\nb'
print(pattern)
```

```python
pattern = r'\d+'
text = 'There are 34 cats, 4 dogs, 5 monkeys'
result = re.findall(pattern, text)
print(result)
```

```python
pattern = 'el'
text = 'hello, what is going on, welcome to the stage'
print(re.findall(pattern, text))
```

```python
pattern = r'\d+'
replacement = 'X'
text = 'Replace 123 and 789 with X'

result = re.sub(pattern, replacement, text)
print(result)
```

```python
a = 'a.b'
b = 'hello hi anb value'
print(re.findall(a, b))
```
