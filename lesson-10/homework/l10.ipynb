```python
def filter_strings_with_substring(strings, substring):
    
    # Define the filtering function
    def contains_substring(string):
        return substring in string

    # Use the filter function to filter out strings with the substring
    filtered_strings = list(filter(contains_substring, strings))

    return filtered_strings

strings = ["Red", "Green", "Orange", "White", "Black", "Pink", "Yellow"]
print("List of words:")
print(strings)
substring = "a"
print("Substring:",substring)
print("Filter out strings with the substring:")
result = filter_strings_with_substring(strings, substring)
print(result)  
```

```python
import itertools
iterable = [1, 4, 9, 6, 7, 12, 10, 14]
a = 10
def sumoftwo(iterable, a):
    for c, d in list(itertools.combinations(iterable[::-1], 2))[::-1]:
        if c+d == a:
            return [c, d]
print(sumoftwo(iterable, a))
#itertools
```

Encapsulation

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self, amount):
        self.balance += amount
        return self.balance
    
    def get_balance(self):
        return self.balance
account = BankAccount('Alice', 100)
account.deposit(50)
print(account.get_balance())
```

Polymorphism

```python
class Bird:
    def fly(self):
        return 'Flies high!'

class Penguin(Bird):
    def fly(self):
        return 'I can\'t fly'
sparrow = Bird()
penguin = Penguin()
print(sparrow.fly())
print(penguin.fly())
```

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model
    def description(self):
        return f'{self.make} {self.model}'
    
class Car(Vehicle):
    def description(self):
        return f'Car: {self.make} {self.model}'
class Truck(Vehicle):
    def description(self):
        return f'Truck: {self.make} {self.model}'

my_car = Car("Toyota", "Corolla")
my_truck = Truck("Ford", "F150")

print(my_car.description())
print(my_truck.description())
```

```python
#Create a class called Book which has attributes title, author, and pages. Include a method called summary to display its information

#Title: The Great Gatsby, Author: F.Scott Fitzgerald, Pages: 180
```

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    def summary(self):
        return f'Title: {self.title}, Author: {self.author}, Pages: {self.pages}'
    
first_book = Book('The Great Gatsby', 'F.Scott Fitzgerald', 180)
print(first_book.summary())
```
